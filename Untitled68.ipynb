```python
import numpy as np


def perceptron(X, y, learning_rate=0.1, epochs=10):


    weights = np.zeros(X.shape[1])
    bias = 0

    for epoch in range(epochs):

        for xi, target in zip(X, y):


            summation = np.dot(xi, weights) + bias


            prediction = 1 if summation >= 0 else 0


            error = target - prediction


            weights += learning_rate * error * xi


            bias += learning_rate * error

    return weights, bias


# Prediction function
def predict(X, weights, bias):

    predictions = []

    for xi in X:

        summation = np.dot(xi, weights) + bias

        prediction = 1 if summation >= 0 else 0

        predictions.append(prediction)

    return predictions



X_and = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_and = np.array([0, 0, 0, 1])

weights_and, bias_and = perceptron(
    X_and,
    y_and
)

pred_and = predict(
    X_and,
    weights_and,
    bias_and
)

print("\nAND OPERATION")
print("Inputs:")
print(X_and)

print("Actual Output:")
print(y_and)

print("Predicted Output:")
print(pred_and)

print("Weights:", weights_and)
print("Bias:", bias_and)




X_or = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_or = np.array([0, 1, 1, 1])

weights_or, bias_or = perceptron(
    X_or,
    y_or
)

pred_or = predict(
    X_or,
    weights_or,
    bias_or
)

print("\nOR OPERATION")
print("Inputs:")
print(X_or)

print("Actual Output:")
print(y_or)

print("Predicted Output:")
print(pred_or)

print("Weights:", weights_or)
print("Bias:", bias_or)




X_not = np.array([
    [0],
    [1]
])

y_not = np.array([1, 0])

weights_not, bias_not = perceptron(
    X_not,
    y_not
)

pred_not = predict(
    X_not,
    weights_not,
    bias_not
)

print("\nNOT OPERATION")
print("Inputs:")
print(X_not.flatten())

print("Actual Output:")
print(y_not)

print("Predicted Output:")
print(pred_not)

print("Weights:", weights_not)
print("Bias:", bias_not)
```
